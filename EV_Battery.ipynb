#EV Battery Quality Classification Using Machine Learning

Objective:
To classify EV battery quality into Grade A, Grade B, and Scrap using manufacturing data.

#Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

##Load Dataset

```python
df = pd.read_csv("ev_battery_qc_data_2026_kaggle.csv")
df.head()
df.info()
df.describe()
```

##Data Cleaning

```python
df = df.drop(columns=[
'Cell_ID',
'Batch_ID',
'Inspector_Comment'
])
```

##Check Missing Values

```python
df.isnull().sum()

df = df.dropna()
```

##Encode Categorical Data

```python
le = LabelEncoder()

df['Production_Line'] = le.fit_transform(df['Production_Line'])
df['Shift'] = le.fit_transform(df['Shift'])
df['Supplier'] = le.fit_transform(df['Supplier'])
df['Defect_Type'] = le.fit_transform(df['Defect_Type'])
df['QC_Grade'] = le.fit_transform(df['QC_Grade'])
```
